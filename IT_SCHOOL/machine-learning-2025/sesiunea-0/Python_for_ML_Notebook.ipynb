## 1. Clase pentru modelarea datelor

```python
class Client:
    def __init__(self, nume, comenzi):
      self.nume = nume
      self.comenzi = comenzi

    def total_comenzi(self):
      return sum(self.comenzi)

    def print_comanda(self):
      return f'Clientul {self.nume} este inregistrat cu comenzile {self.comenzi}'

ana = Client("Ana", [250, 90])
mihai = Client("Mihai", [130, 310, 500])

print(ana.total_comenzi())
```

```python
# Pornind de la instantele deja definite - ana

mia = ana  # aliasing
mia.comenzi.append(60)
print("Mia:", mia.print_comanda())
print("Ana:", ana.print_comanda())
```

```python
# Shallow copy -> face o copie noua a listei, dar elementele nu sunt copii separate (doar container-ul se schimba)
import copy

# 1) Obiectul initial
print("Ana original:", ana.print_comanda())
# are comenzile [250, 90, 60]

# 2) Shallow copy
shallow = copy.copy(ana)
print("Shallow copy initial:", shallow.print_comanda())

# 3) Modificam numele (atribut imutabil) -> se copiaza separat
shallow.nume = "Anisoara"

print("\nDupa reasignare shallow.nume:")
print("  Ana original:", ana.print_comanda())
print("  Shallow copy:", shallow.print_comanda())

# 4) Modificam lista (atribut mutabil) -> ramane shared
shallow.comenzi.append(300)
print("\nDupa append(300) în shallow.orders:")
print("  Ana original:", ana.print_comanda())
print("  Shallow copy:", shallow.print_comanda())
```

```python
# Deep copy -> copiaza recursiv si elementele din interior, deci totul devine independent

import copy
deep_copy = copy.deepcopy(ana)  # deep copy
deep_copy.comenzi.append(20)
print("Ana original:", ana.print_comanda())
print("Ana deep copy:", deep_copy.print_comanda())
```

```python
# Mostenire -> VIPClient mosteneste clasa Client => are tot ce stie Client (metode, atribute) dar poate adauga / modifica metode

class VIPClient(Client):
    def total_comenzi(self):
        return super().total_comenzi() * 0.9  # reducere 10%

vip = VIPClient('Vlad', [100, 200])
print("Total VIP cu reducere:", vip.total_comenzi())
```

## 7. Functional Programming

```python
comenzi_ana = [100, 250, 90]
rezultat = list(map(lambda x: x * 2, comenzi_ana))
print(rezultat)
```

```python
from functools import reduce

# reduce() -> REDUCE la o singura valoare
comenzi = [100, 250, 90]
total = reduce(lambda x, y: x + y, comenzi)
print(total)

# filter() -> pastreaza doar elementele pentru care functia returneaza TRUE
comenzi = [100, 250, 90]
mari = list(filter(lambda x: x > 100, comenzi))
print(mari)
```

## 11. Clean Code

```python
# Principiul 1. Nume clare si semnificative

# Exemplu prost
a = [1, 2, 3]
for x in a:
    print(x)

# Exemplu bun
comenzi = [100, 200, 300]
for comanda in comenzi:
    print(f"Comanda: {comanda}")
```

```python
# Principiul 2. Functii scurte, o singura responsabilitate

# Exemplu prost
def proceseaza(client):
    total = sum(client.comenzi)
    print(f"{client.nume} are comenzi de {total} lei")
    if total > 500:
        print("Client VIP")

# Exemplu bun
def calculeaza_total(client):
    return sum(client.comenzi)

def afiseaza_total(client):
    total = calculeaza_total(client)
    print(f"{client.nume} are comenzi de {total} lei")

def verifica_vip(client):
    if calculeaza_total(client) > 500:
        print("Client VIP")
```

```python
# Principiul 3. Comentarii utile, dar nu in exces

# Exemplu prost
# Iterăm prin fiecare element din listă
for comanda in comenzi:
    print(comanda)

# Exemplu bun
# Filtrăm comenzile suspecte (sub 0) înainte de a calcula totalul
comenzi_filtrate = [c for c in comenzi if c >= 0]
```

```python
# Principiul 4. Evitarea duplicarii de cod

# Exemplu prost
# Calculez totalul comenzilor doar pentru Ana și Mihai
total_ana = sum([c for c in clienti["Ana"].comenzi if c > 0])
total_mihai = sum([c for c in clienti["Mihai"].comenzi if c > 0])

# Exemplu bun
def total_valid_comenzi(client):
    return sum([c for c in client.comenzi if c > 0])

total_ana = total_valid_comenzi(clienti["Ana"])
total_mihai = total_valid_comenzi(clienti["Mihai"])
```

# 12. Concluzii - putem folosi tot ce am invatat pentru a realiza un magazin online

```python
import json
from functools import reduce

# 1. Clasă pentru Client
class Client:
    def __init__(self, nume):
        self.nume = nume
        self.comenzi = []

    def adauga_comanda(self, valoare):
        try:
            valoare = float(valoare)
            if valoare <= 0:
                raise ValueError("Comanda trebuie să fie pozitivă.")
            self.comenzi.append(valoare)
        except ValueError:
            print(f"[Eroare] Valoare invalidă pentru {self.nume}: {valoare}")

    def total(self):
        return sum(self.comenzi)

    def comenzi_mari(self, prag=200):
        return list(filter(lambda x: x > prag, self.comenzi))

# 2. Cream clienți și adăugăm comenzi
ana = Client("Ana")
mihai = Client("Mihai")

comenzi_ana = ["250", "abc", "90", "-10"]
comenzi_mihai = [130, 310, 500]

for c in comenzi_ana:
    ana.adauga_comanda(c)

for c in comenzi_mihai:
    mihai.adauga_comanda(c)

# 3. Dicționar de clienți
clienti = {"Ana": ana, "Mihai": mihai}

# 4. Totaluri pe client (dict comprehension)
totaluri = {nume: client.total() for nume, client in clienti.items()}

# 5. Găsim comenzile mari pentru fiecare client
comenzi_mari = {nume: client.comenzi_mari() for nume, client in clienti.items()}

# 6. Scriem în JSON
with open("clienti.json", "w") as f:
    json.dump(clienti, f, default=lambda o: o.__dict__, indent=2)

# 7. Afișare rezultate (cu clean code)
def afiseaza_rezumat(client):
    print(f"Client: {client.nume}")
    print(f" - Comenzi: {client.comenzi}")
    print(f" - Total: {client.total()} lei")
    print(f" - Comenzi mari: {client.comenzi_mari()}")
    print()

for client in clienti.values():
    afiseaza_rezumat(client)
```
